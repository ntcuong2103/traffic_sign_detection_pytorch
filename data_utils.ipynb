```python
import pandas as pd

annotations = pd.read_csv('za_traffic_2020/traffic_train/annotation.csv')


```

```python
import numpy as np
bboxes_array = np.array([bbox.strip('[]').split(', ') for bbox in annotations['bbox'].values[:]], dtype=int)
```

```python
MIN_SIZE = 30
```

```python
bboxes_array[(bboxes_array[:, 2] < MIN_SIZE) * (bboxes_array[:, 3] < MIN_SIZE)].shape
```

```python
bboxes_array_filter = bboxes_array[(bboxes_array[:, 2] >= MIN_SIZE) + (bboxes_array[:, 3] >= MIN_SIZE)]
```

```python
from pprint import pprint
def analyses(bboxes_array):
    pprint({'min':bboxes_array[:, 2:4].min(0), 
            'max':bboxes_array[:, 2:4].max(0), 
            'max_ratio':(bboxes_array[:, 2] / bboxes_array[:, 3]).max(), 
            'min_ratio':(bboxes_array[:, 2] / bboxes_array[:, 3]).min()})
```

```python
analyses(bboxes_array)
```

```python
analyses(bboxes_array_filter)
```
